```python
import numpy as np
import pandas as pd

import tensorflow as tf
from tensorflow import keras
from keras import layers
from keras.layers import SimpleRNN, Embedding,Dense,LSTM,GRU, Bidirectional

from keras.models import Sequential
from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences
from keras.datasets import imdb
```

```python
from keras.datasets import imdb
(X_train, y_train), (X_test, y_test)=imdb.load_data(num_words=10000)
```

```python
# Flatten all sequences into one list
all_words = [word for review in X_train for word in review]

# Find unique word indices
unique_words = len(set(all_words))
unique_words
```

```python
X_train=pad_sequences(X_train, maxlen=100)
X_test=pad_sequences(X_test, maxlen=100)
```

**DEEP RNN**

```python
model=Sequential([
   Embedding(unique_words+1,32,input_shape=(100,)) ,
   SimpleRNN(5, return_sequences=True),
   SimpleRNN(5),
   Dense(1, activation="sigmoid")

])
```

```python
model.summary()
```

```python
# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history = model.fit(X_train, y_train, epochs=1, batch_size=32, validation_split=0.2)
```

**Deep LSTM**

```python
model=Sequential([
   Embedding(unique_words+1,32,input_shape=(100,)) ,
   LSTM(5, return_sequences=True),
   LSTM(5),
   Dense(1, activation="sigmoid")

])
model.summary()
```

**Deep GRU**

```python
model=Sequential([
   Embedding(10000,32,input_shape=(100,)) ,
   GRU(5, return_sequences=True),
   GRU(5),
   Dense(1, activation="sigmoid")

])
model.summary()
```

**Bidrecrtional RNN**

```python
model=Sequential([
   Embedding(10000,32,input_shape=(100,)) ,
   Bidirectional(SimpleRNN(5)),
   Dense(1, activation="sigmoid")

])
model.summary()
```

**BiDirection LSTM**

```python

model=Sequential([
   Embedding(10000,32,input_shape=(100,)) ,
   Bidirectional(LSTM(5)),
   Dense(1, activation="sigmoid")

])
model.summary()
```

**BiDirection GRU**

```python
model=Sequential([
   Embedding(10000,32,input_shape=(100,)) ,
   Bidirectional(GRU(5)),
   Dense(1, activation="sigmoid")

])
model.summary()
```
