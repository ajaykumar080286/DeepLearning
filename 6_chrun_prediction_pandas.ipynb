```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler, OrdinalEncoder,OneHotEncoder
from sklearn.metrics import accuracy_score

from sklearn.compose import ColumnTransformer
```

```python
df=pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/DeepLearning/master/Churn_Modelling.csv")
```

```python
df.head()
```

```python
df.drop(columns=["RowNumber","CustomerId","Surname"],inplace=True)
```

```python
df.head()
```

```python
df=pd.get_dummies(df,columns=["Geography","Gender"],drop_first=True)
```

```python
df.shape
```

```python
X = df.drop(columns=['Exited'])
y = df['Exited'].values
```

```python
X.shape
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=0)
```

```python
scale=StandardScaler()
X_train_trf=scale.fit_transform(X_train)
X_test_trf=scale.transform(X_test)
```

```python
X_train_trf
```

```python
import tensorflow as tf
from tensorflow import keras
from keras import Sequential
from keras.layers import Dense
```

```python
model=Sequential()

model.add(Dense(11,activation="relu",input_dim=11))
model.add(Dense(11,activation="relu"))
model.add(Dense(1,activation="sigmoid"))
```

```python
model.summary()
```

```python
model.compile(optimizer="Adam",loss="binary_crossentropy", metrics=["accuracy"])
```

```python
model.fit(X_train_trf,y_train,epochs=10,validation_split=0.2)
```

```python
y_pred=model.predict(X_test_trf)
```

```python
y_pred=np.where(y_pred>0.5,1,0)
```
