```python
import pandas as pd
import tensorflow

from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, SimpleRNN
```

```python
model=Sequential()

model.add(SimpleRNN(3,input_shape=(4,5)))
model.add(Dense(1,activation="sigmoid"))
```

```python
model.summary()
```
