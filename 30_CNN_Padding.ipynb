```python
import pandas as pd
import numpy as np

import tensorflow
from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense,Conv2D,Dropout, Flatten
from keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
X_test.shape
```

```python
X_train.shape
```

**Without Padding**

```python
model=Sequential()
model.add(Conv2D(32, kernel_size=(3,3), activation="relu",padding="valid", input_shape=(28,28,1)))
model.add(Conv2D(32, kernel_size=(3,3), padding="valid",activation="relu"))
model.add(Conv2D(32, kernel_size=(3,3), padding="valid",activation="relu"))

model.add(Flatten())

model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

**With Padding**

```python
model=Sequential()
model.add(Conv2D(32, kernel_size=(3,3), activation="relu",padding="same", input_shape=(28,28,1)))
model.add(Conv2D(32, kernel_size=(3,3), padding="same",activation="relu"))
model.add(Conv2D(32, kernel_size=(3,3), padding="same",activation="relu"))

model.add(Flatten())

model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

**Strides With Padding**

```python
model=Sequential()
model.add(Conv2D(32, kernel_size=(3,3), activation="relu",padding="same",strides=(2,2), input_shape=(28,28,1)))
model.add(Conv2D(32, kernel_size=(3,3), padding="same",strides=(2,2),activation="relu"))
model.add(Conv2D(32, kernel_size=(3,3), padding="same",strides=(2,2),activation="relu"))

model.add(Flatten())

model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

***Strides Without Padding***

```python
model=Sequential()
model.add(Conv2D(32, kernel_size=(3,3), activation="relu",padding="valid",strides=(2,2), input_shape=(28,28,1)))
model.add(Conv2D(32, kernel_size=(3,3), padding="valid",strides=(2,2),activation="relu"))
model.add(Conv2D(32, kernel_size=(3,3), padding="valid",strides=(2,2),activation="relu"))

model.add(Flatten())

model.add(Dense(128,activation="relu"))
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```
