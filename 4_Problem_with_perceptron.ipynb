```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.linear_model import Perceptron
from mlxtend.plotting import plot_decision_regions
```

```python
or_data=pd.DataFrame({
    "input1": [1,1,0,0],
    "input2": [1,0,1,0],
    "output": [1,1,1,0]
})
```

```python
or_data
```

```python
and_data=pd.DataFrame()
and_data['input1']=[1,1,0,0]
and_data['input2']=[1,0,1,0]
and_data['output']=[1,0,0,0]
```

```python
and_data
```

```python
xor_data=pd.DataFrame()
xor_data['input1']=[1,1,0,0]
xor_data['input2']=[1,0,1,0]
xor_data['output']=[0,1,1,0]
```

```python
xor_data
```

```python
sns.scatterplot(x=and_data["input1"], y=and_data["input2"], hue=and_data["output"])
```

```python
sns.scatterplot(x=or_data["input1"], y=or_data["input2"], hue=or_data["output"])
```

```python
sns.scatterplot(x=xor_data["input1"], y=xor_data["input2"], hue=xor_data["output"])
```

```python
clf1=Perceptron()
clf2=Perceptron()
clf3=Perceptron()
```

```python
clf1.fit(and_data[["input1", "input2"]].values, and_data["output"].values)
coef_=clf1.coef_
inter_=clf1.intercept_

m=-(coef_[0][0]/coef_[0][1])
b=-(inter_[0]/coef_[0][1])

x=np.linspace(-1,1,4)
y=m*x+b

plt.plot(x,y)
sns.scatterplot(x=and_data["input1"], y=and_data["input2"], hue=and_data["output"])
```

```python
clf2.fit(or_data[["input1", "input2"]].values, or_data["output"].values)
coef_=clf2.coef_
inter_=clf2.intercept_

m=-(coef_[0][0]/coef_[0][1])
b=-(inter_[0]/coef_[0][1])

x=np.linspace(-1,1,4)
y=m*x+b

plt.plot(x,y)
sns.scatterplot(x=or_data["input1"], y=or_data["input2"], hue=or_data["output"])
```

```python
clf3.fit(xor_data[["input1", "input2"]].values, xor_data["output"].values)
coef_=clf3.coef_
inter_=clf3.intercept_

m=-(coef_[0][0]/coef_[0][1])
b=-(inter_[0]/coef_[0][1])

x=np.linspace(-1,1,4)
y=m*x+b

plt.plot(x,y)
sns.scatterplot(x=xor_data["input1"], y=xor_data["input2"], hue=xor_data["output"])
```

```python
plot_decision_regions(xor_data[["input1", "input2"]].values,xor_data["output"].values,clf=clf3)
```
