```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

from sklearn.linear_model import Perceptron
```

```python
df= pd.read_csv("https://raw.githubusercontent.com/ajaykumar080286/DeepLearning/master/placement.csv")
```

```python
df.head()
```

```python
X=df.iloc[:,0:2]
y=df.iloc[:,-1]
```

```python
sns.scatterplot(x=df["cgpa"], y=df["resume_score"], hue=df["placed"])
```

```python
ps=Perceptron()
```

```python
ps.fit(X,y)
```

```python
coef_=ps.coef_
```

```python
coef_
```

```python
inter=ps.intercept_
inter
```

```python
from mlxtend.plotting import  plot_decision_regions
```

```python
plot_decision_regions(X.values, y.values, clf=ps,legend=2)
```

```python
coef_[0][1]
```

```python
m=-(coef_[0][0]/coef_[0][1])
m
```

```python
b=-(inter/coef_[0][1])
b
```

```python
X_input=np.linspace(5,10,100)
y_imput=m*X_input+b
```

```python
plt.figure(figsize=(10,6))
plt.scatter(X.iloc[:,0],X.iloc[:,1],c=y,cmap='winter',s=100)
plt.plot(X_input,y_imput,color='red',linewidth=3)
```
