**Wikipedia Retriever**

```python
from langchain_community.retrievers import WikipediaRetriever
```

```python
retriever=WikipediaRetriever(top_k_results=2,lang="en")
```

```python
query="is TCS is IT company"
```

```python
docs= retriever.invoke(query)
```

```python
for i,doc in enumerate(docs):
  print("-----Result-----",i+1)
  print("\n")
  print(doc)
```
