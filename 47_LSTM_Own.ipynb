```python
import pandas as pd
import numpy as np
import tensorflow as tf

from tensorflow import keras
from keras.models import Sequential
from keras.layers import Dense, LSTM, Embedding

from tensorflow.keras.preprocessing.text import Tokenizer
from tensorflow.keras.preprocessing.sequence import pad_sequences

from keras.utils import to_categorical
```

```python
feqs="""my name is ajay
I am from Bihar"""
```

```python
feqs
```

```python
tokenizer=Tokenizer()
```

```python
tokenizer.fit_on_texts([feqs])
```

```python
tokenizer.word_index
```

```python
input_sequences=[]
for sentence in feqs.split("\n"):
    seq_text=tokenizer.texts_to_sequences([sentence])[0]

    for i in range(1,len(seq_text)):
      input_sequences.append(seq_text[:i+1])

```

```python
input_sequences
```

```python
length=[len(x) for x in input_sequences]
```

```python
max_length=max(length)
```

```python
max_length
```

```python
padded_input_sequences=pad_sequences(input_sequences, maxlen=max_length, padding="pre")
```

```python
padded_input_sequences
```

```python
X=padded_input_sequences[:,:-1]
y=padded_input_sequences[:,-1]
```

```python
X
```

```python
y
```

```python
X.shape
```

```python
len(tokenizer.word_index)
```

```python
y.shape
```

```python
y=to_categorical(y, num_classes=len(tokenizer.word_index)+1)
```

```python
y
```

```python
model=Sequential()
model.add(Embedding(len(tokenizer.word_index)+1, 4, input_length=X.shape[1]))
model.add(LSTM(150))
model.add(Dense(len(tokenizer.word_index)+1, activation="softmax"))
```

```python
model.summary()
```

```python
model.compile(loss="categorical_crossentropy", optimizer="adam", metrics=["accuracy"])
```

```python
model.fit(X,y,epochs=100)
```

```python
import time
text = "i"

for i in range(3):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=3, padding='pre')
  # predict
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
      time.sleep(2)
```
